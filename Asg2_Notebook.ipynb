## Test 7 - Nearest-Neighbor Prediction
**Expected** Predict the label of the closest historical patient using only the nine features. The TumorType value should be ignored.  

**Input**
Historical data from test_load.csv.

Patient 1 (same as historical row 11, which is benign) - expect 0
| Feature1 | Feature2 | Feature3 | Feature4 | Feature5 | Feature6 | Feature7 | Feature8 | Feature9 |
|----------|----------|----------|----------|----------|----------|----------|----------|----------|
| 2 | 1 | 1 | 1 | 2 | 1 | 2 | 1 | 1 |

Patient 2 (same as the malignant row) - expect 1
| Feature1 | Feature2 | Feature3 | Feature4 | Feature5 | Feature6 | Feature7 | Feature8 | Feature9 |
|----------|----------|----------|----------|----------|----------|----------|----------|----------|
| 8 | 9 | 10 | 8 | 7 | 10 | 9 | 7 | 1 |

```python
predictTest = DataTable("data/test_load.csv")
nnTest = NearestNeighborPredictor(predictTest)

def tumorName(value):
    if value == 0.0:
        return "Benign"
    return "Malignant"

exactPatient = {"Feature1": 2, "Feature2": 1, "Feature3": 1, "Feature4": 1, "Feature5": 2, "Feature6": 1, "Feature7": 2, "Feature8": 1, "Feature9": 1}
malignantPatient = {"Feature1": 8, "Feature2": 9, "Feature3": 10, "Feature4": 8, "Feature5": 7, "Feature6": 10, "Feature7": 9, "Feature8": 7, "Feature9": 1}

if nnTest.distance(exactPatient, predictTest.get_row(11)) == 0:
    print("Patient 1 matches historical row 11 exactly: Yes") #Same as historical row 11 so this should be Yes
else:
    print("Patient 1 matches historical row 11 exactly: No")
print("Patient 1 prediction:", tumorName(nnTest.predict(exactPatient))) #Should be Benign
print("Patient 2 prediction:", tumorName(nnTest.predict(malignantPatient))) #Closest to the malignant row so this should be Malignant

malignantPatient["TumorType"] = 0.0
print("Patient 3 prediction:", tumorName(nnTest.predict(malignantPatient))) #Wrong TumorType is added but it should be ignored, still Malignant
```

## Test 8 - Out-of-Range Feature
**Expected** Raise an OutOfSampleError with a meaningful message when a feature is below the historical min or above the historical max. A value exactly equal to the min or max is allowed.  

**Input**
Historical data from test_load.csv. The range for Feature1 is [1, 8] and the range for Feature4 is [1, 8].

| Test | Feature1 | Feature2 | Feature3 | Feature4 | Feature5 | Feature6 | Feature7 | Feature8 | Feature9 | Expected |
|------|----------|----------|----------|----------|----------|----------|----------|----------|----------|----------|
| Above max | 2 | 1 | 1 | **9** | 2 | 1 | 2 | 1 | 1 | Error |
| Below min | **0** | 1 | 1 | 1 | 2 | 1 | 2 | 1 | 1 | Error |
| On the max | 2 | 1 | 1 | **8** | 2 | 1 | 2 | 1 | 1 | No error, prediction is Benign |

```python
predictTest = DataTable("data/test_load.csv")
modeTest = MajorityClassPredictor(predictTest)
nnTest = NearestNeighborPredictor(predictTest)

def tumorName(value):
    if value == 0.0:
        return "Benign"
    return "Malignant"

aboveMax = {"Feature1": 2, "Feature2": 1, "Feature3": 1, "Feature4": 9, "Feature5": 2, "Feature6": 1, "Feature7": 2, "Feature8": 1, "Feature9": 1}
belowMin = {"Feature1": 0, "Feature2": 1, "Feature3": 1, "Feature4": 1, "Feature5": 2, "Feature6": 1, "Feature7": 2, "Feature8": 1, "Feature9": 1}
onMax = {"Feature1": 2, "Feature2": 1, "Feature3": 1, "Feature4": 8, "Feature5": 2, "Feature6": 1, "Feature7": 2, "Feature8": 1, "Feature9": 1}

try:
    modeTest.predict(aboveMax)
except OutOfSampleError as e:
    print("Above max:", e) #Feature4 is above the max of 8

try:
    nnTest.predict(belowMin)
except OutOfSampleError as e:
    print("Below min:", e) #Feature1 is below the min of 1

try:
    print("On the max:", tumorName(nnTest.predict(onMax))) #Feature4 is exactly the max of 8 so there should be no error, prints Benign
except OutOfSampleError as e:
    print("On the max:", e)
```

## Test 10 - Accuracy Calculation
**Expected** Calculate the accuracy of both predictors against the 100 labels. Predictions that are None (out of sample) are skipped.  

**Input**
Small check: predictions [Benign, Malignant, Malignant, None] vs actual [Benign, Malignant, Benign, Benign]. 2 are correct out of the 3 that count, so the accuracy should be about 66.67%.

Full check: 599 historical rows, 100 new cases and 100 labels. The mode predictor always guesses Benign, so its accuracy should equal the percentage of benign cases in the labels.

```python
print("Small check:", calculate_accuracy([0.0, 1.0, 1.0, None], [0.0, 1.0, 0.0, 0.0])) #2 correct out of 3 (None is skipped), so about 66.67%

historical.calc_missing_val()
new_cases.calc_missing_val(historical)

modePredictor = MajorityClassPredictor(historical)
nnPredictor = NearestNeighborPredictor(historical)

modePredictions = []
nnPredictions = []

for i in range(new_cases.get_num_rows()):
    patient = new_cases.get_row(i)

    try:
        modePredictions.append(modePredictor.predict(patient))
    except OutOfSampleError as e:
        modePredictions.append(None)
        print(f"Case {i} (mode): {e}")

    try:
        nnPredictions.append(nnPredictor.predict(patient))
    except OutOfSampleError as e:
        nnPredictions.append(None)
        print(f"Case {i} (Nearest Neighbor): {e}")

actualLabels = labels.get_column("TumorType")
print("Number of labels:", len(actualLabels)) #Should be 100 labels
print("Benign labels:", actualLabels.count(0.0)) #79 are benign so the mode accuracy should be 79%
print("Mode accuracy:", calculate_accuracy(modePredictions, actualLabels))
print("Nearest Neighbor accuracy:", calculate_accuracy(nnPredictions, actualLabels))
```
